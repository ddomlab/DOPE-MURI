```python
import re
from pathlib import Path
import pandas as pd
from rdkit import Chem, RDLogger

RDLogger.DisableLog("rdApp.*")

kraken = pd.read_csv(Path("gp_hub") / "reference" / "kraken" / "kraken_identifiers.csv")

# Catalysts whose SMILES has the ligand missing or drawn wrong; used only when
# the SMILES itself finds no Kraken ligand.
NAME_OVERRIDES = {
    "pd(oac)2-2g-xphos": 1,      # XPhos -- SMILES is plain Pd(OAc)2
    "davephos-pd-g3": 263,       # DavePhos -- SMILES has NHiPr, not NMe2
    "aphos-pd-g3": 216,          # AmPhos -- SMILES has NH2, not NMe2
    "ph-sphos-pd(allyl)": 7,     # PhSPhos -- SMILES has P meta, not ortho
    "qphos-pd-g3": 330,          # QPhos -- ferrocene/phenyls missing from SMILES
    "pd(qphos)2": 330,
    "pd(oac)2-qphos": 330,
}

METALS = {"Pd", "Ni"}  # Fe stays: ferrocenyl ligands keep it in Kraken


def no_stereo_smiles(mol):
    mol = Chem.Mol(mol)
    Chem.RemoveStereochemistry(mol)
    return Chem.MolToSmiles(mol)


def phosphine_ligands(smiles):
    """Canonical SMILES of every P-containing fragment once Pd/Ni are removed."""
    mol = Chem.MolFromSmiles(smiles, sanitize=False) if isinstance(smiles, str) else None
    if mol is None:
        return []
    rw = Chem.RWMol(mol)
    for idx in sorted((a.GetIdx() for a in rw.GetAtoms() if a.GetSymbol() in METALS), reverse=True):
        rw.RemoveAtom(idx)

    ligands = []
    for frag in Chem.GetMolFrags(rw.GetMol(), asMols=True, sanitizeFrags=False):
        if not any(a.GetSymbol() == "P" for a in frag.GetAtoms()):
            continue
        # Undo the precatalyst bonding: neutral atoms, dative bonds -> single
        for a in frag.GetAtoms():
            a.SetFormalCharge(0)
            a.SetNumExplicitHs(0)
            a.SetNoImplicit(False)
        for b in frag.GetBonds():
            if b.GetBondType() == Chem.BondType.DATIVE:
                b.SetBondType(Chem.BondType.SINGLE)
        try:
            Chem.SanitizeMol(frag)
            ligands.append(no_stereo_smiles(frag))
        except Exception:
            ligands.append(None)
    return ligands


# Kraken lookup: stereo-free canonical SMILES -> id
kraken_by_smiles = {}
for smi, kid in zip(kraken["can_smiles"], kraken["id"]):
    mol = Chem.MolFromSmiles(smi)
    if mol is not None:
        kraken_by_smiles.setdefault(no_stereo_smiles(mol), int(kid))
kraken_name = dict(zip(kraken["id"].astype(int), kraken["ligand"]))


def kraken_id(name, smiles):
    ids = {kraken_by_smiles.get(lig) for lig in phosphine_ligands(smiles)}
    # One Kraken ligand and nothing else phosphorus-containing (e.g. a bidentate
    # Xantphos next to PPh3 would make the assignment ambiguous)
    if len(ids) == 1 and None not in ids:
        return ids.pop()
    if not ids:
        return NAME_OVERRIDES.get(name, pd.NA)
    return pd.NA


pairs = df_reagents_no_additive[["Catalyst Name", "Catalyst SMILES"]].drop_duplicates()
pair_ids = {(n, s): kraken_id(n, s) for n, s in pairs.itertuples(index=False)}

df_reagents_no_additive["Kraken ID"] = pd.array(
    [pair_ids[(n, s)] for n, s in zip(df_reagents_no_additive["Catalyst Name"],
                                      df_reagents_no_additive["Catalyst SMILES"])],
    dtype="Int64",
)

# ---- Precatalyst type, read from the SMILES ----
AMINOBIPHENYL = Chem.MolFromSmarts("[#7]~[#6]1~[#6](~[#6]2~[#6]~[#6]~[#6]~[#6]~[#6]~2)~[#6]~[#6]~[#6]~[#6]~1")
PHENETHYLAMINE = Chem.MolFromSmarts("[#7;!a]~[#6;!a]~[#6;!a]~[#6]1~[#6]~[#6]~[#6]~[#6]~[#6]~1")
OTHER_PALLADACYCLE = Chem.MolFromSmarts("[#7;!a]~[#6;!a]~[#6]~[#6]-[Pd]")
ALLYL = Chem.MolFromSmarts("[#6;!R]~[#6;!R]~[#6;!R]-[Pd]")
NHC_PD = Chem.MolFromSmarts("[#7]~[#6](~[Pd])~[#7]")
ARYL_PD = Chem.MolFromSmarts("[c;-1,$(c-[Pd])]")
ACETATE = Chem.MolFromSmarts("[#6;D1]-[#6](=[#8])-[#8]")
DBA = Chem.MolFromSmarts("[#8]=[#6](-[#6]=[#6]-c)-[#6]=[#6]-c")


def backbone_match(mol, pattern):
    """First match of pattern that is not part of a phosphine ligand."""
    for match in mol.GetSubstructMatches(pattern):
        if not any(nb.GetSymbol() == "P" for i in match for nb in mol.GetAtomWithIdx(i).GetNeighbors()):
            return match
    return None


def precatalyst_type(smiles):
    mol = Chem.MolFromSmiles(smiles, sanitize=False) if isinstance(smiles, str) else None
    if mol is None:
        return pd.NA
    mol.UpdatePropertyCache(strict=False)
    Chem.FastFindRings(mol)
    elements = {a.GetSymbol() for a in mol.GetAtoms()}

    if "Ni" in elements:
        return "Ni(II) complex" if "P" in elements else "None: in situ Ni salt"
    if "Pd" not in elements:
        return "Unknown: no metal in SMILES"

    match = backbone_match(mol, AMINOBIPHENYL)
    if match:
        n_methyl = any(nb.GetSymbol() == "C" and nb.GetDegree() == 1
                       for nb in mol.GetAtomWithIdx(match[0]).GetNeighbors())
        if "S" in elements:          # mesylate / triflate
            return "G4" if n_methyl else "G3"
        if "Cl" in elements:
            return "G2"
    if backbone_match(mol, PHENETHYLAMINE):
        return "G1"
    if mol.HasSubstructMatch(NHC_PD):
        return "NHC-Pd(allyl)"
    if mol.HasSubstructMatch(ALLYL):
        return "Allyl"
    if mol.HasSubstructMatch(OTHER_PALLADACYCLE):
        return "Palladacycle (other)"
    if mol.HasSubstructMatch(DBA):
        return "None: in situ Pd(dba)"
    if mol.HasSubstructMatch(ACETATE):
        return "None: in situ Pd(OAc)2"
    if mol.HasSubstructMatch(ARYL_PD):
        return "Oxidative addition complex"
    if elements & {"Cl", "Br", "I"}:
        return "PdX2L2"
    return "Pd(0)Ln"


def precatalyst_type_from_name(name):
    """The type the catalyst name claims, or None if the name doesn't say."""
    n = str(name).lower()
    if "pd(oac)2" in n:
        return "None: in situ Pd(OAc)2"
    if "palladacycle" in n:
        return "G1"
    for gen in "1234":
        if re.search(rf"(^|-)(g{gen}|{gen}g)(-|$)", n):
            return f"G{gen}"
    if "(allyl)" in n:
        return "Allyl"
    return None


# Ligand disagreements found by comparing each name with its SMILES by hand
LIGAND_MISMATCH = {
    "pd(oac)2-2g-xphos": "SMILES has no ligand (plain Pd(OAc)2)",
    "davephos-pd-g3": "SMILES ligand has NHiPr, DavePhos is NMe2",
    "aphos-pd-g3": "SMILES ligand has NH2, AmPhos is NMe2",
    "ph-sphos-pd(allyl)": "SMILES ligand has PPh2 meta, PhSPhos is ortho",
    "qphos-pd-g3": "SMILES QPhos is missing the ferrocene/phenyls",
    "pd(qphos)2": "SMILES QPhos is missing the ferrocene/phenyls",
    "pd(oac)2-qphos": "SMILES QPhos is missing the ferrocene/phenyls",
    "xphos-pd(allyl)": "SMILES ligand is Xantphos, not XPhos",
    "alphos-pd(allyl)": "SMILES ligand is MorDalPhos, not AlPhos",
    "pd(oac)2-tbubinap": "SMILES ligand is TrixiePhos, not tBuBINAP",
}


def mismatch_reason(name, smiles_type):
    reasons = []
    if name in LIGAND_MISMATCH:
        reasons.append(LIGAND_MISMATCH[name])
    claimed = precatalyst_type_from_name(name)
    if claimed is not None and claimed != smiles_type:
        reasons.append(f"name says {claimed}, SMILES is {smiles_type}")
    return "; ".join(reasons) if reasons else pd.NA


pair_types = {(n, s): precatalyst_type(s) for n, s in pairs.itertuples(index=False)}
pair_reasons = {(n, s): mismatch_reason(n, pair_types[(n, s)]) for n, s in pairs.itertuples(index=False)}

keys = list(zip(df_reagents_no_additive["Catalyst Name"], df_reagents_no_additive["Catalyst SMILES"]))
df_reagents_no_additive["Precatalyst Type"] = [pair_types[k] for k in keys]
df_reagents_no_additive["Mismatch Reason"] = [pair_reasons[k] for k in keys]
df_reagents_no_additive["Name/SMILES Mismatch"] = df_reagents_no_additive["Mismatch Reason"].notna()

# ---- Summary ----
has_id = df_reagents_no_additive["Kraken ID"].notna()
print(f"Rows with a Kraken ID: {has_id.sum():,} / {len(has_id):,}")
print(f"Unique Kraken ligands: {df_reagents_no_additive['Kraken ID'].nunique()}")
print(f"Rows tagged Name/SMILES Mismatch: {df_reagents_no_additive['Name/SMILES Mismatch'].sum():,}\n")
print(df_reagents_no_additive["Precatalyst Type"].value_counts().to_string(), "\n")

display(df_reagents_no_additive[df_reagents_no_additive["Name/SMILES Mismatch"]]
        .groupby(["Catalyst Name", "Precatalyst Type", "Mismatch Reason"])
        .size().reset_index(name="rows"))

summary = (df_reagents_no_additive
           .assign(**{"Kraken Ligand": df_reagents_no_additive["Kraken ID"].map(kraken_name)})
           .groupby(["Catalyst Name", "Precatalyst Type", "Kraken ID", "Kraken Ligand"], dropna=False)
           .size().reset_index(name="rows"))
display(summary)
df_cleaned_reagents = df_reagents_no_additive[~df_reagents_no_additive["Name/SMILES Mismatch"]]
```

```python
import textwrap
import matplotlib.pyplot as plt
from matplotlib.ticker import StrMethodFormatter, AutoMinorLocator
from nispo import smiles_to_iupac

FONT_SIZE = 15          # ordinary category labels
LUMPED_FONT_SIZE = 18   # the pooled "Other" segments start here
LUMPED_MIN_SIZE = 9     # ...and shrink rather than lose their (#)
NAME_MIN_SIZE = 7       # floor for a name label
COUNT_MIN_SIZE = 5      # floor for a number-only label
EXCLUDE = {"Yield", "Product SMILES"}
COUNT_ONLY = {"Aryl SMILES": 100, "Amine SMILES": 100}  # feature -> min reactions to get its own segment
LUMPED_PREFIX = "Cat. ("   # shorthand for "categories"; keep the is_other tests in sync


def top_categories(series, max_n=12, min_n=3, min_frac=0.015):
    """Keep categories holding at least min_frac of the rows (between min_n and
    max_n of them) and lump the rest into 'Other'. min_frac keeps every segment
    wide enough to hold a label."""
    counts = series.value_counts(dropna=False)
    frac = counts / counts.sum()
    n_keep = min(max(min_n, int((frac >= min_frac).sum())), max_n, len(counts))
    if len(counts) - n_keep == 1:   # an "Other (1)" is just the last category
        n_keep += 1
    kept, rest = counts.iloc[:n_keep], counts.iloc[n_keep:]
    if len(rest) > 0:
        kept = pd.concat([kept, pd.Series({f"{LUMPED_PREFIX}{len(rest)})": rest.sum()})])
    return kept


def keep_at_least(series, min_count):
    """Every category with at least min_count rows; the rest lumped into 'Other'."""
    counts = series.value_counts(dropna=False)
    kept, rest = counts[counts >= min_count], counts[counts < min_count]
    if len(rest) > 0:
        kept = pd.concat([kept, pd.Series({
            f"{len(rest)} categories <{min_count}:": rest.sum()
        })])
    return kept


def label_for(feature, value):
    # Both lumped labels: "Cat. (n)" from top_categories and
    # "n categories <min:" from keep_at_least.
    if str(value).startswith(LUMPED_PREFIX) or "categories <" in str(value):
        return str(value)
    if pd.isna(value):
        return "Not in Kraken" if feature == "Kraken ID" else "NA"
    if feature == "Kraken ID":
        return kraken_name.get(int(value), str(value))
    if feature.endswith("SMILES"):
        name = smiles_to_iupac(value)
        return name if name != "" else value
    return str(value)


def _clip(name, n):
    return name if len(name) <= n else name[:n - 1] + "…"


def _wrap(name, width):
    lines = textwrap.wrap(name, width) or [name]
    if len(lines) > 3:
        lines = lines[:3]
        lines[-1] = lines[-1][:-1] + "…"
    return lines


def label_modes(name, count, lumped=False):
    """Label modes from most to least complete; each a list of (text, rotation).

    Inside a mode the variants compete only on font size, so the arrangement
    that renders largest wins -- a segment flips to vertical exactly when that
    buys a bigger label.

    Rotated name+count is written INLINE ("name  count"). A second line would be
    laid out beside the name in the rotated frame, which reads as the number
    sitting below it.
    """
    count_s = f"{count:,}"
    if name is None:            # count-only segment
        yield [(count_s, 0), (count_s, 90)]
        return
    if lumped:
        # Never truncate a lumped label -- the (#) is the point of it. Only the
        # orientation and the font size may change.
        yield [(f"{name}\n{count_s}", 0), (f"{name}  {count_s}", 90)]
        yield [(name, 0), (name, 90)]
        yield [(count_s, 0), (count_s, 90)]
        return

    both = [("\n".join(_wrap(name, w) + [count_s]), 0)
            for w in (len(name), 30, 24, 18, 14, 10)]
    both += [(f"{_clip(name, n)}  {count_s}", 90)
             for n in (len(name), 26, 20, 15, 11, 7)]
    yield both
    yield ([(_clip(name, n), 0) for n in (len(name), 22, 16, 11)]
           + [(_clip(name, n), 90) for n in (len(name), 26, 20, 15, 11, 7, 4)])
    yield [(count_s, 0), (count_s, 90)]


def fit_label(text, rect, name, count, renderer, lumped=False):
    """Most complete label that fits, at the largest size any arrangement allows.

    Modes are the outer loop and sizes the inner one, so the font shrinks all the
    way down before a mode is abandoned: a small name+count beats a large number
    on its own.
    """
    box = rect.get_window_extent(renderer)
    if lumped:
        sizes = range(LUMPED_FONT_SIZE, LUMPED_MIN_SIZE - 1, -1)
    elif name is None:
        sizes = range(FONT_SIZE, COUNT_MIN_SIZE - 1, -1)
    else:
        sizes = range(FONT_SIZE, NAME_MIN_SIZE - 1, -1)
    pad = 2 if name is None else 4
    for variants in label_modes(name, count, lumped):
        for size in sizes:
            text.set_fontsize(size)
            for candidate, rotation in variants:
                text.set_text(candidate)
                text.set_rotation(rotation)
                ext = text.get_window_extent(renderer)
                if ext.width <= box.width - pad and ext.height <= box.height - pad:
                    return
    text.set_text("")


df_plot = df_cleaned_reagents
features = [c for c in df_plot.columns if c not in EXCLUDE and df_plot[c].notna().any()]
total = len(df_plot)
colors = plt.get_cmap("tab20").colors

fig, ax = plt.subplots(figsize=(24, 2.1 * len(features) + 1))
segments = []   # (rect, text, name, count, is_other)

for row, feature in enumerate(features):
    left = 0
    if feature in COUNT_ONLY:
        counts = keep_at_least(df_plot[feature], COUNT_ONLY[feature])
    else:
        counts = top_categories(df_plot[feature])
    for i, (value, count) in enumerate(counts.items()):
        is_other = str(value).startswith(LUMPED_PREFIX) or "categories <" in str(value)
        if feature in COUNT_ONLY and not is_other:
            name = None   # show the number only
        else:
            name = label_for(feature, value)
        color = "lightgrey" if is_other else colors[i % len(colors)]
        rect = ax.barh(row, count, left=left, height=0.85, color=color,
                       edgecolor="white", linewidth=1)[0]
        text = ax.text(left + count / 2, row, "", ha="center", va="center",
                       fontsize=FONT_SIZE, rotation_mode="anchor")
        segments.append((rect, text, name, count, is_other))
        left += count

ax.set_yticks(range(len(features)), labels=features, fontsize=16)
ax.set_ylim(len(features) - 0.5, -0.5)  # first feature at the top
ax.set_xlim(0, total)
ax.set_xlabel("Number of reactions", fontsize=16)
ax.set_title(f"Feature distributions (n = {total:,})", fontsize=18, pad=12)
ax.xaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))
ax.xaxis.set_minor_locator(AutoMinorLocator(2))
ax.tick_params(axis="x", which="both", direction="in", labelsize=13)
ax.tick_params(axis="y", length=0)
for side in ("top", "right"):
    ax.spines[side].set_visible(False)

# Fit labels after the layout is final, so segment sizes are known
fig.tight_layout()
fig.canvas.draw()
renderer = fig.canvas.get_renderer()
for rect, text, name, count, is_other in segments:
    fit_label(text, rect, name, count, renderer, lumped=is_other)

plt.show()
```
